```python
# Data Preprocessing and Data Loading
IMG_SIZE = 224
BATCH_SIZE = 32
NUM_WORKERS = 4

# Training Data Preprocessing
train_transform = transforms.Compose([

    # Resize all images to 224 × 224 pixels
    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    # Randomly flip images horizontally
    transforms.RandomHorizontalFlip(p=0.5),

    # Randomly flip images vertically
    transforms.RandomVerticalFlip(p=0.5),

    # Randomly rotate images up to ±30°
    transforms.RandomRotation(degrees=30),

    # Randomly adjust brightness, contrast,
    # saturation, and hue
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.15,
        hue=0.05
    ),

    # Randomly translate the image by up to 5%
    transforms.RandomAffine(
        degrees=0,
        translate=(0.05, 0.05)
    ),

    # Convert image into PyTorch tensor
    transforms.ToTensor(),

    # Normalize pixel values using ImageNet statistics
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),

    # Randomly erase a small region of the image
    # to improve model robustness
    transforms.RandomErasing(
        p=0.2,
        scale=(0.02, 0.1)
    )
])

# Validation and Test Data Preprocessing
val_test_transform = transforms.Compose([

    # Resize images
    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    # Convert to tensor
    transforms.ToTensor(),

    # Normalize pixel values
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# Load Dataset
train_dataset = ImageFolder(
    TRAIN_PATH,
    transform=train_transform
)

val_dataset = ImageFolder(
    VAL_PATH,
    transform=val_test_transform
)

test_dataset = ImageFolder(
    TEST_PATH,
    transform=val_test_transform
)

CLASS_NAMES = train_dataset.classes
NUM_CLASSES = len(CLASS_NAMES)

print(f"Classes       : {CLASS_NAMES}")
print(f"Class Mapping : {train_dataset.class_to_idx}")

# Weighted Random Sampler (Balances class sampling during training)
train_labels = [label for _, label in train_dataset.samples]

class_counts = Counter(train_labels)

class_weights = {
    cls: 1.0 / count
    for cls, count in class_counts.items()
}

sample_weights = [
    class_weights[label]
    for label in train_labels
]

sampler = WeightedRandomSampler(
    sample_weights,
    len(sample_weights)
)

# Create DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=NUM_WORKERS,
    pin_memory=False
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=False
)

# Dataset Summary
print("\n✅ DataLoaders created successfully!")

print(f"Training Images   : {len(train_dataset):5d} | {len(train_loader)} batches")
print(f"Validation Images : {len(val_dataset):5d} | {len(val_loader)} batches")
print(f"Testing Images    : {len(test_dataset):5d} | {len(test_loader)} batches")
```

```python
# -------------------------------------------------------
# Histopathology-Pretrained ResNet-50
# TCGA-BRCA SimCLR pretrained model
# Regularized Version
# -------------------------------------------------------

import torch
import torch.nn as nn
import timm


class ResNet50Classifier(nn.Module):

    def __init__(self, num_classes=2):
        super().__init__()

        # ---------------------------------------------------
        # Load ResNet-50 pretrained on histopathology images
        # ---------------------------------------------------

        self.backbone = timm.create_model(
            "hf-hub:1aurent/resnet50.tcga_brca_simclr",
            pretrained=True,
            num_classes=0
        )

        # ---------------------------------------------------
        # Freeze all backbone layers
        # ---------------------------------------------------

        for param in self.backbone.parameters():
            param.requires_grad = False

        # ---------------------------------------------------
        # Unfreeze only Layer4
        # ---------------------------------------------------

        for param in self.backbone.layer4.parameters():
            param.requires_grad = True

        # ---------------------------------------------------
        # ResNet-50 feature dimension
        # ---------------------------------------------------

        in_features = self.backbone.num_features

        print("Backbone feature dimension:", in_features)

        # ---------------------------------------------------
        # Regularized Custom Classifier
        # ---------------------------------------------------

        self.classifier = nn.Sequential(

            # Reduced classifier size
            nn.Linear(in_features, 256),

            nn.BatchNorm1d(256),

            nn.ReLU(inplace=True),

            # Stronger dropout
            nn.Dropout(0.6),

            # Reduced hidden layer
            nn.Linear(256, 64),

            nn.ReLU(inplace=True),

            # Stronger dropout
            nn.Dropout(0.5),

            # Final classification layer
            nn.Linear(64, num_classes)
        )

        # ---------------------------------------------------
        # Train classifier
        # ---------------------------------------------------

        for param in self.classifier.parameters():
            param.requires_grad = True

    def forward(self, x):

        # Extract pathology features
        features = self.backbone(x)

        # Classification
        output = self.classifier(features)

        return output


# -------------------------------------------------------
# Test Model
# -------------------------------------------------------

dummy = torch.randn(2, 3, 224, 224)

model = ResNet50Classifier(
    num_classes=NUM_CLASSES
)

output = model(dummy)


# -------------------------------------------------------
# Parameter Count
# -------------------------------------------------------

trainable = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

total = sum(
    p.numel()
    for p in model.parameters()
)


print("\n" + "=" * 60)
print("Histopathology-Pretrained ResNet-50")
print("=" * 60)

print(f"Output Shape         : {output.shape}")
print(f"Trainable Parameters : {trainable:,}")
print(f"Total Parameters     : {total:,}")
print("=" * 60)
```

```python
def train_model(model, train_loader, val_loader, model_name, num_epochs=30):

    # Move model to CPU/GPU
    model = model.to(DEVICE)

    # ------------------------------------------
    # Loss Function (Class Weighted)
    # ------------------------------------------

    class_weights = [1.0 / class_counts[i] for i in range(NUM_CLASSES)]

    criterion = nn.CrossEntropyLoss(
        weight=torch.FloatTensor(class_weights).to(DEVICE)
    )

    # ------------------------------------------
    # Optimizer (Recommended)
    # ------------------------------------------

    optimizer = optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=1e-5,
        weight_decay=1e-3
    )

    # ------------------------------------------
    # Learning Rate Scheduler
    # ------------------------------------------

    scheduler = ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=3
    )

    # ------------------------------------------
    # Early Stopping
    # ------------------------------------------

    early_stop_patience = 7
    early_stop_counter = 0

    best_validation_loss = float("inf")
    best_validation_accuracy = 0.0
    best_model_weights = None

    # ------------------------------------------
    # Store Training History
    # ------------------------------------------

    history = {
        "train_loss": [],
        "train_accuracy": [],
        "validation_loss": [],
        "validation_accuracy": []
    }

    print("=" * 70)
    print(f"Training Model : {model_name}")
    print("=" * 70)

    # ==========================================
    # Training Loop
    # ==========================================

    for epoch in range(num_epochs):

        # -----------------------------
        # Training
        # -----------------------------

        model.train()

        running_loss = 0.0
        correct_predictions = 0
        total_samples = 0

        for images, labels in tqdm(
            train_loader,
            desc=f"Epoch {epoch+1}/{num_epochs}",
            leave=False
        ):

            images = images.to(DEVICE)
            labels = labels.to(DEVICE)

            optimizer.zero_grad()

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            # Gradient Clipping
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)

            optimizer.step()

            running_loss += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)

            correct_predictions += (predictions == labels).sum().item()

            total_samples += labels.size(0)

        train_loss = running_loss / total_samples
        train_accuracy = correct_predictions / total_samples

        # -----------------------------
        # Validation
        # -----------------------------

        model.eval()

        validation_loss = 0.0
        validation_correct = 0
        validation_samples = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(DEVICE)
                labels = labels.to(DEVICE)

                outputs = model(images)

                loss = criterion(outputs, labels)

                validation_loss += loss.item() * images.size(0)

                predictions = outputs.argmax(dim=1)

                validation_correct += (predictions == labels).sum().item()

                validation_samples += labels.size(0)

        validation_loss /= validation_samples
        validation_accuracy = validation_correct / validation_samples

        # ------------------------------------------
        # Save Training History
        # ------------------------------------------

        history["train_loss"].append(train_loss)
        history["train_accuracy"].append(train_accuracy)
        history["validation_loss"].append(validation_loss)
        history["validation_accuracy"].append(validation_accuracy)

        # ------------------------------------------
        # Update Learning Rate
        # ------------------------------------------

        scheduler.step(validation_loss)

        # ------------------------------------------
        # Save Best Model
        # ------------------------------------------

        if validation_loss < best_validation_loss:

            best_validation_loss = validation_loss
            best_validation_accuracy = validation_accuracy

            best_model_weights = copy.deepcopy(model.state_dict())

            torch.save(
                best_model_weights,
                os.path.join(
                    SAVE_DIR,
                    f"best_{model_name.replace(' ','_')}.pth"
                )
            )

            early_stop_counter = 0

        else:

            early_stop_counter += 1

        # ------------------------------------------
        # Print Epoch Result
        # ------------------------------------------

        print(
            f"Epoch [{epoch+1:02d}/{num_epochs}] | "
            f"Train Loss: {train_loss:.4f} | "
            f"Train Acc: {train_accuracy:.4f} | "
            f"Val Loss: {validation_loss:.4f} | "
            f"Val Acc: {validation_accuracy:.4f}"
        )

        # ------------------------------------------
        # Early Stopping
        # ------------------------------------------

        if early_stop_counter >= early_stop_patience:

            print("\nEarly stopping triggered.")
            break

    print("\nTraining Completed Successfully!")
    print(f"Best Validation Accuracy : {best_validation_accuracy:.4f}")
    print(f"Best Validation Loss     : {best_validation_loss:.4f}")

    # Load Best Model

    if best_model_weights is not None:
        model.load_state_dict(best_model_weights)

    return model, history


print("Training function is ready.")
```

```python
def evaluate_model(model, test_loader, model_name):

    # Set model to evaluation mode
    model.eval()

    all_predictions = []
    all_labels = []
    all_probabilities = []

    # Disable gradient computation during evaluation
    with torch.no_grad():

        for images, labels in test_loader:

            images = images.to(DEVICE)

            outputs = model(images)

            probabilities = torch.softmax(outputs, dim=1)

            predictions = outputs.argmax(dim=1)

            all_predictions.extend(predictions.cpu().numpy())
            all_labels.extend(labels.numpy())
            all_probabilities.extend(probabilities.cpu().numpy())

    # Convert lists to NumPy arrays
    all_predictions = np.array(all_predictions)
    all_labels = np.array(all_labels)
    all_probabilities = np.array(all_probabilities)

    # Calculate Classification Metrics
    report = classification_report(
        all_labels,
        all_predictions,
        target_names=CLASS_NAMES,
        output_dict=True
    )

    # Calculate ROC-AUC Score (Binary Classification)
    auc_score = None

    if NUM_CLASSES == 2:
        auc_score = roc_auc_score(
            all_labels,
            all_probabilities[:, 1]
        )

    # Print Evaluation Results
    print("\n" + "=" * 70)
    print(f"Evaluation Results : {model_name}")
    print("=" * 70)

    print(classification_report(
        all_labels,
        all_predictions,
        target_names=CLASS_NAMES
    ))

    if auc_score is not None:
        print(f"AUC-ROC Score : {auc_score:.4f}")


    # Confusion Matrix
    cm = confusion_matrix(all_labels, all_predictions)

    plt.figure(figsize=(6, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=CLASS_NAMES,
        yticklabels=CLASS_NAMES
    )

    plt.title(
        f"Confusion Matrix - {model_name}",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel("Predicted Class")
    plt.ylabel("True Class")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            SAVE_DIR,
            f"confusion_matrix_{model_name.replace(' ', '_').lower()}.png"
        ),
        dpi=150
    )

    plt.show()

    # Return Evaluation Metrics
    results = {
        "accuracy": report["accuracy"],
        "precision": report["macro avg"]["precision"],
        "recall": report["macro avg"]["recall"],
        "f1_score": report["macro avg"]["f1-score"],
        "auc_roc": auc_score,
        "probabilities": all_probabilities,
        "true_labels": all_labels
    }

    return results


print("Evaluation function is ready.")
```

```python
NUM_EPOCHS = 30
results    = {}
```

```python
print("Training ResNet-50 Model...")

# Train the model
model_resnet50, history_resnet50 = train_model(
    ResNet50Classifier(num_classes=NUM_CLASSES),
    train_loader,
    val_loader,
    model_name="ResNet50",
    num_epochs=NUM_EPOCHS
)

# Save the trained model
torch.save(
    model_resnet50.state_dict(),
    os.path.join(SAVE_DIR, "model_resnet50_TCGA_BRCA.pth")
)

# Evaluate the model
results["ResNet50"] = evaluate_model(
    model_resnet50,
    test_loader,
    model_name="ResNet50"
)

# Save training history
results["ResNet50"]["history"] = history_resnet50

print("\n✅ ResNet-50 training and evaluation completed successfully!")
print(f"💾 Model saved to: {os.path.join(SAVE_DIR, 'model_resnet50_TCGA_BRCA.pth')}")
```
